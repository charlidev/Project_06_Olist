```sql
-- 1. Permisos a nivel Catálogo
GRANT USE CATALOG ON CATALOG olist_dev TO `data_engineers`;
GRANT USE CATALOG ON CATALOG olist_dev TO `bi_analysts`;
GRANT USE CATALOG ON CATALOG olist_dev TO `data_scientists`;


-- 2. INGENIEROS DE DATOS (Ana y Tú)

-- Permiso total en las tres capas.
GRANT USE SCHEMA, SELECT, MODIFY, CREATE TABLE ON SCHEMA olist_dev.bronze TO `data_engineers`;
GRANT USE SCHEMA, SELECT, MODIFY, CREATE TABLE ON SCHEMA olist_dev.silver TO `data_engineers`;
GRANT USE SCHEMA, SELECT, MODIFY, CREATE TABLE ON SCHEMA olist_dev.gold TO `data_engineers`;


-- 3. ANALISTAS DE BI (Beto)

-- Solo se otorga acceso a Gold. Al no hacer un GRANT en Bronze o Silver, el acceso está denegado por defecto.
GRANT USE SCHEMA, SELECT ON SCHEMA olist_dev.gold TO `bi_analysts`;


-- 4. CIENTÍFICOS DE DATOS (Carlos)

-- Solo se otorga lectura a Silver y Gold. Al no otorgar MODIFY ni CREATE, no pueden escribir.
GRANT USE SCHEMA, SELECT ON SCHEMA olist_dev.silver TO `data_scientists`;
GRANT USE SCHEMA, SELECT ON SCHEMA olist_dev.gold TO `data_scientists`;
```
